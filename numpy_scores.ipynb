```python
#..................................
#Task 1
#..................................
import numpy as np

np.random.seed(42)
scores = np.random.randint(50, 100, size = (5, 4))
print("scores :\n", scores)

#The score of the 3rd student in the 2nd subject
print ("Score of 3rd student in  the second subject:")
print(scores [2, 1])


#All scores of the last 2 students (all subjects)
print("All scores of the last 2 students (all subjects):")
print(scores [-2,1])

#All scores for the first 3 students in subjects 2 and 3 only
print("All scores for the first 3 students in subjects 2 and 3 only")
print(scores [:3, 1:3])



```

```python
# -------------------------------
# Task 2 — Broadcasting & Analysis
# -------------------------------


# 1 Column-wise mean (average per subject)
column_mean = np.round(scores.mean(axis=0), 2)
print("\n column-wise mean:", column_mean)

# 2️ Add curve using broadcasting'
curve = np.array([5, 3, 7, 2])
curved_scores = scores + curve
curved_scores = np.clip(curved_scores, None, 100)
print("\n Curved scores are :\n", curved_scores)


# 3️ Row-wise maximum (best subject per student)
row_max = curved_scores.max(axis=1)
print("\n Row wise max of best student per subject:", row_max)
```

```python
# ----------------------------
# Task 3 — Normalization & Identification
# ----------------------------

# Min-max normalization per row
row_min = curved_scores.min(axis=1, keepdims=True)
row_max = curved_scores.max(axis=1, keepdims=True)

normalized = (curved_scores - row_min) / (row_max - row_min)

print("\nNormalized Scores:\n", normalized)

# 1️ Index of single highest value in normalized array
max_index = np.unravel_index(np.argmax(normalized), normalized.shape)

print("\nHighest normalized value at:")
print("Student index:", max_index[0])
print("Subject index:", max_index[1])

# 2️ All curved scores strictly above 90
above_90 = curved_scores[curved_scores > 90]

print("\nScores above 90:", above_90)
```
